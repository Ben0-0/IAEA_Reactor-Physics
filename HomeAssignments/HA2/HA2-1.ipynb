$\epsilon$ fast fission factor: extra neutrons from fast fission; it’s the total neutron count after fast fission divided by the thermal fission neutrons.\
$p$ resonance escape probability: the fraction of neutrons that slow down past the resonance energies without being captured (mostly by U-238).\
$f$ thermal utilization factor: the fraction of thermal neutrons absorbed in the fuel rather than in the moderator.\
$\eta$ reproduction factor: neutrons produced per thermal neutron absorbed in the fuel.

So first, $\epsilon$ is calculated by: $\epsilon = \frac{\text{total number of fission neutrons}}{\text{number of fission neutrons from just thermal fissions}}$

Next, the resonance escape probability p, which is calculated by: $p = \frac{\text{fission neutrons slowed to thermal energies without absorption}}{\text{total fission neutrons}}$

Next, the thermal utilization factor, which is calculated $f = \frac{\text{thermal neutrons absorbed by the fuel isotope}}{\text{thermal neutrons absorbed anywhere}}$

Last the reproduction factor, $\eta$, is calculated by: $\eta = \frac{\text{neutrons produced from thermal fissions}}{\text{thermal absorption in fuel isotope}}$ 

```python
epsilon = 1026/1000
p = 930/1026
f = 858/930
eta = 1000/858
k_inf = epsilon*p*f*eta
print(k_inf)
# The infinite  multiplication factor k_inf is close to 1; thus this reactor is in the critical state with a stable neutron economy & power production.
```
